# XP Crafter spike (spec 5.5) on Colab
Plain PPO only, seed 0, 1,000,000 steps per area; reads no RND result. Use a **GPU runtime** (T4 is fine) if you can; CPU works too.
The spike script is not on GitHub until pushed, so cell 3 uploads it from your laptop (`scripts/spikes/crafter_spike.py`). The script does **not resume**: after a disconnect, rerun that area's cell (it restarts from step 0).
Results go to Drive under `xp-results/crafter-spike/area{A}-seed0/` (`curve.csv`, `summary.json`). Download that folder into `results/crafter-spike/` in the repo, then run the Task 5 analysis locally.
